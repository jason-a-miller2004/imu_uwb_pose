```python
import os
import os.path as osp
import pickle
from imu_uwb_pose.eval_metrics import get_trans_metrics

results_dir = osp.relpath("./data/results/LOO_trans_5e-3")

files = os.listdir(results_dir)
errs = 0
for file in files:
  fpath = osp.join(results_dir, file)
  with open(fpath, 'rb') as f:
    outputs = pickle.load(f)
    one_sec_err = get_trans_metrics(outputs)
    break
    errs += one_sec_err
    print(f'processing {file} err: {one_sec_err}')

errs /= len(files)
print(f'average err {errs}')
```
